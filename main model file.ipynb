```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import warnings
warnings.filterwarnings("ignore")

import kagglehub
path = kagglehub.dataset_download("arshid/iris-flower-dataset")
```

```python
import os

# Dynamically search for the CSV file in the downloaded path

files = [f for f in os.listdir(path) if f.endswith('.csv')]
if files:
    csv_file_path = os.path.join(path, files[0])
    df = pd.read_csv(csv_file_path)
    print(f"Successfully loaded: {csv_file_path}")
    display(df.head())
else:
    print(f"No CSV files found in {path}. Available files: {os.listdir(path)}")
```

```python
# split the target
X = df.drop(columns=['species'])
y = df['species']
```

```python
# split the test data
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=50,
    stratify=y
)
```

```python
# mini-max sscaling

from sklearn.preprocessing import MinMaxScaler

minmax = MinMaxScaler()

X_train_minmax = minmax.fit_transform(X_train)
X_test_minmax = minmax.transform(X_test)

#######################################################################
# z-score scaling

#from sklearn.preprocessing import StandardScaler

#zscore = StandardScaler()

#X_train_zscore = zscore.fit_transform(X_train)
#X_test_zscore = zscore.transform(X_test)

######################################################################

print("Min-Max Scaling:")
print(X_train_minmax[:5])

#print("\nZ-score Standardization:")
#print(X_train_zscore[:5])
```

```python
# use KNN to traning dataset with [k=3,5,7]
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

for k in [3, 5, 7]:

    knn = KNeighborsClassifier(n_neighbors=k)

    knn.fit(X_train_minmax, y_train)

    y_pred = knn.predict(X_test_minmax)

    accuracy = accuracy_score(y_test, y_pred)

    print(f"K = {k}, Accuracy = {accuracy * 100:.2f}%")
```
